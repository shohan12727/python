```python
def spam_detector(comment: str) -> bool:
  spam_keywords = [
        "make a lot of money",
        "buy now",
        "click this"
    ]

  comment = comment.lower()

  for keyWords in spam_keywords:
    if keyWords in comment:
      return True

  return False

comments = [
    "You can make a lot of money from home!",
    "This is a normal discussion comment",
    "Buy now and get 50% off",
    "Click this link to continue"
]

for c in comments:
    print(f"'{c}' -> {'SPAM' if spam_detector(c) else 'NOT SPAM'}")
```
